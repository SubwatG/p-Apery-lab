```python
# Pure Python implementation — no SageMath needed!
# Generating function 1/∏(1-x^{a_i}) coefficients = number of representations via DP

def generating_function_coefficients(a, d, k, num_terms):
    """
    Coefficients of 1/((1-x^a)(1-x^{a+d})...(1-x^{a+(k-1)d}))
    using DP coin change (fast, no SageMath required).
    """
    A = [a + i*d for i in range(k)]
    dp = [0] * num_terms
    dp[0] = 1
    for gen in A:
        for n in range(gen, num_terms):
            dp[n] += dp[n - gen]
    return dp

def generating_function_coefficients_A(A, num_terms):
    """Same DP for arbitrary generating set A."""
    dp = [0] * num_terms
    dp[0] = 1
    for gen in A:
        for n in range(gen, num_terms):
            dp[n] += dp[n - gen]
    return dp

def gf(D, p):
    """p-Frobenius number: largest n with at most p representations."""
    for i in range(len(D) - 1, -1, -1):
        if D[i] <= p:
            return i
    print('Please try again with a longer list of D')
    return -1

def numerical(D, p):
    """S_{>p} : all integers with at least p+1 representations."""
    return [i for i, x in enumerate(D) if x > p]

def p_Apery(D, p):
    """p-Apéry set: minimal n in each residue class mod a with >p reps."""
    return [i for i, x in enumerate(D)
            if (p == 0 and x > p and i - a < 0) or (x > p and D[i - a] <= p)]

def diop_two_condition(a, b, d):
    """Solve a*x + b*y = d with y < a."""
    if any(param <= 0 for param in [a, b]) or d < 0:
        return "Error: a, b, must be positive, d non-negative."
    solutions = []
    for x in range(d // a + 1):
        remaining = d - a * x
        if remaining % b == 0:
            y = remaining // b
            if y < a:
                solutions.append((x, y))
    return solutions

def diop(a, b, c, d):
    """Solve a*x + b*y + c*z = d."""
    if any(param <= 0 for param in [a, b, c]) or d < 0:
        return "Error: a, b, c positive, d non-negative."
    solutions = []
    for x in range(d // a + 1):
        remaining_after_x = d - a * x
        for y in range(remaining_after_x // b + 1):
            remaining_after_y = remaining_after_x - b * y
            if remaining_after_y % c == 0:
                z = remaining_after_y // c
                solutions.append((x, y, z))
    return solutions
```

# p-Apery set for A.P.

```python
# Uses functions from Cell 1
# Compute p-Apéry set for an arithmetic progression A = [a, a+d, ..., a+(k-1)d]

a, d, k = 5, 2, 3
A = [a + d*i for i in range(k)]
print(f'A = {A}')

num_terms = 10000
D = generating_function_coefficients(a, d, k, num_terms)

print('\n')

pAp_before = [0 for i in range(a)]
for p in range(0, 41):
    print(f'p = {p}')

    pAp = p_Apery(D, p)
    sorted_pAp = sorted(pAp, key=lambda x: x % a)
    print(f'{p}-Ap = {sorted_pAp}')

    # Difference from previous p-Apéry set (componentwise)
    dif_pAp = [sorted_pAp[i] - pAp_before[i] for i in range(a)]
    print(f'     dif = {dif_pAp}')
    pAp_before = sorted_pAp

    # Diophantine representation: n = x*A[1] + y*A[2] conditionally
    pt = [diop_two_condition(A[1], A[2], i) for i in sorted_pAp]
    print(f'     (x,y) = {pt}')
    print('')
```

```python
import matplotlib.pyplot as plt

# Uses generating_function_coefficients, p_Apery, diop_two_condition from Cell 1
# Plot (x,y) representations of p-Apéry elements for each p

a, d, k = 5, 2, 3
A = [a + d*i for i in range(k)]
print(f'A = {A}')
num_terms = 10000

D = generating_function_coefficients(a, d, k, num_terms)

for p in range(0, 10):
    print(f'\np = {p}')
    pAp = p_Apery(D, p)
    print(f'  {p}-Apéry set = {pAp}')

    pt = [diop_two_condition(A[1], A[2], i) for i in pAp]
    print(f'  (x,y) reps = {pt}')

    for idx, solutions in enumerate(pt):
        for x, y in solutions:
            plt.scatter(x, y, color='blue')
            plt.text(x + 0.2, y, f'{p}', fontsize=8)

    # Highlight the maximal element (last in sorted pAp) in red
    if pt and pt[-1]:
        for x, y in pt[-1]:
            plt.scatter(x, y, color='red')
            plt.text(x + 0.2, y, f'{p}', fontsize=10, color='red', fontweight='bold')

plt.title('p-Apéry Points (x,y) with Labels')
plt.xlabel('x (coefficient of a+d)')
plt.ylabel('y (coefficient of a+2d)')
plt.grid(True)
plt.show()
```

# import p-Ap set into Excel files

```python
# Functions are defined in Cell 1 — this cell is kept for reference
# generating_function_coefficients uses pure Python DP (no SageMath)
# p_Apery computes p-Apéry sets using coefficient array D
# diop_two_condition solves a*x + b*y = d with constraint y < a
```

```python
import pandas as pd
from prettytable import PrettyTable
from math import gcd

# Uses generating_function_coefficients, p_Apery from Cell 1
# Export p-Apéry sets for multiple (a,d) pairs to Excel files

k = 3  # number of terms in AP
for a in range(5, 6):
    for d in range(1, 10):
        if gcd(a, d) != 1:
            continue
        A = [a + d*i for i in range(k)]
        print(A)
        num_terms = 20000 if d < 15 else 40000

        D = generating_function_coefficients(a, d, k, num_terms)

        table = PrettyTable()
        table.field_names = ["p"] + [f"pt[{i}]" for i in range(a)]
        data = []

        for p in range(501):
            pAp = p_Apery(D, p)
            sorted_pAp = sorted(pAp, key=lambda x: x % a)
            pt = sorted_pAp          # store actual p-Apéry values, not (x,y) pairs
            pt += [""] * (a - len(pt))
            table.add_row([p] + pt)
            data.append([p] + pt)

        df = pd.DataFrame(data, columns=table.field_names)
        excel_filename = f"p_apery_table_a={a}_d={d}.xlsx"
        df.to_excel(excel_filename, index=False)
        print(f"  → {excel_filename} saved")
```
